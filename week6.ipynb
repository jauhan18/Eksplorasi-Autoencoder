```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# === Load Data ===
(x_train, _), (x_test, _) = tf.keras.datasets.fashion_mnist.load_data()
x_train = x_train.astype("float32") / 255.
x_test = x_test.astype("float32") / 255.
x_train = x_train.reshape((len(x_train), 28, 28, 1))
x_test = x_test.reshape((len(x_test), 28, 28, 1))

# === Build Model ===
input_img = layers.Input(shape=(28,28,1))

# ----- Encoder (3 lapis) -----
x = layers.Conv2D(32, (3,3), activation="relu", padding="same")(input_img)   # 28x28x32
x = layers.MaxPooling2D((2,2), padding="same")(x)                           # 14x14x32

x = layers.Conv2D(64, (3,3), activation="relu", padding="same")(x)          # 14x14x64
x = layers.MaxPooling2D((2,2), padding="same")(x)                           # 7x7x64

x = layers.Conv2D(128, (3,3), activation="relu", padding="same")(x)         # 7x7x128
latent = layers.MaxPooling2D((2,2), padding="same")(x)                      # 4x4x128 (latent space)

# ----- Decoder (3 lapis) -----
x = layers.Conv2D(128, (3,3), activation="relu", padding="same")(latent)    # 4x4x128
x = layers.UpSampling2D((2,2))(x)                                           # 8x8x128

x = layers.Conv2D(64, (3,3), activation="relu", padding="same")(x)          # 8x8x64
x = layers.UpSampling2D((2,2))(x)                                           # 16x16x64

x = layers.Conv2D(32, (3,3), activation="relu", padding="same")(x)          # 16x16x32
x = layers.UpSampling2D((2,2))(x)                                           # 32x32x32

decoded = layers.Conv2D(1, (3,3), activation="sigmoid", padding="same")(x)  # 32x32x1

# Crop ke (28x28)
decoded = layers.Cropping2D(((2,2),(2,2)))(decoded)                         # 28x28x1

# === Compile Model ===
autoencoder = models.Model(input_img, decoded)
autoencoder.compile(optimizer="adam", loss="binary_crossentropy")

autoencoder.summary()

# === Training ===
autoencoder.fit(x_train, x_train,
                epochs=20,
                batch_size=256,
                shuffle=True,
                validation_data=(x_test, x_test))

# === Reconstruction Test ===
decoded_imgs = autoencoder.predict(x_test)

n = 10
plt.figure(figsize=(20, 4))
for i in range(n):
    # Original
    ax = plt.subplot(2, n, i+1)
    plt.imshow(x_test[i].reshape(28,28), cmap="gray")
    plt.title("Original")
    plt.axis("off")

    # Reconstructed
    ax = plt.subplot(2, n, i+1+n)
    plt.imshow(decoded_imgs[i].reshape(28,28), cmap="gray")
    plt.title("Reconstructed")
    plt.axis("off")
plt.show()
```

# Analisis Kuantitatif Rekonstruksi

```python
from sklearn.metrics import mean_squared_error
from skimage.metrics import structural_similarity as ssim

# Ambil label asli (untuk analisis per kelas)
(_, y_train), (_, y_test) = tf.keras.datasets.fashion_mnist.load_data()

# Ambil rekonstruksi
decoded_imgs = autoencoder.predict(x_test)

# Hitung error per sampel
mse_scores = []
ssim_scores = []

for i in range(len(x_test)):
    orig = x_test[i].reshape(28,28)
    recon = decoded_imgs[i].reshape(28,28)
    
    # MSE
    mse_val = mean_squared_error(orig.flatten(), recon.flatten())
    mse_scores.append(mse_val)
    
    # SSIM
    ssim_val = ssim(orig, recon, data_range=1.0)
    ssim_scores.append(ssim_val)

mse_scores = np.array(mse_scores)
ssim_scores = np.array(ssim_scores)

print("Rata-rata MSE:", mse_scores.mean())
print("Rata-rata SSIM:", ssim_scores.mean())
```

# Analisis Per Kelas Fashion MNIST

```python
import pandas as pd

class_names = ["T-shirt/Top","Trouser","Pullover","Dress","Coat",
               "Sandal","Shirt","Sneaker","Bag","Ankle boot"]

results = []

for cls in range(10):
    idxs = np.where(y_test == cls)[0]
    mse_mean = mse_scores[idxs].mean()
    ssim_mean = ssim_scores[idxs].mean()
    results.append([class_names[cls], mse_mean, ssim_mean])

df_results = pd.DataFrame(results, columns=["Class","MSE","SSIM"])
print(df_results)
```
